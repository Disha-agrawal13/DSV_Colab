```python
from IPython import get_ipython
from IPython.display import display
# %%
#ques 1 = create a csv with no header and 5 columns . load it with your own column names and make the first column as index column.
import pandas as pd
data = [
    [1, "Disha", "A", 85, 90],
    [2, "Bhoomi", "B", 78, 88],
    [3, "Muskan", "A", 92, 95],
    [4, "Ananya", "C", 80, 84],
    [5, "Shivi", "B", 89, 91]
]
df = pd.DataFrame(data)
df.to_csv("students.csv", header=False, index=False)
df = pd.read_csv("students.csv", header=None, index_col=0)
print(df)
# %%
# ques 2 write a file using ; as delimiter and -999 for missing values. read it so -999 becomes  nan .
data = {
    "Name": ["Disha", "Bhoomi", "Muskan", "Ananya"],
    "Marks": [85, -999, 92, 78],
    "Age": [19, 20, -999, 18]
}
df = pd.DataFrame(data)
df.to_csv("data.csv", sep=";", index=False)
df = pd.read_csv("data.csv", sep=";", na_values=-999)
print(df)
# %%
#ques 3 read a big csv in chunks of 2000 rows and compute the mean of one column per key across all chunks .
data = {
    "Name": ["Disha", "Bhoomi", "Muskan", "Ananya", "Shivi"] * 1000,
    "Marks": [85, 90, 78, 88, 92] * 1000
}
df = pd.DataFrame(data)
df.to_csv("bigdata.csv", index=False)
print("File created!")
total = 0
count = 0
for chunk in pd.read_csv("bigdata.csv", chunksize=2000):
    total += chunk["Marks"].sum()
    count += chunk["Marks"].count()
mean = total / count
print("Mean =", mean)
# %%
# ques 4 export a dataframe with orient = "records". read it back with pd.read_json , and check equality with  .equals
df = pd.DataFrame({
    "Name": ["Disha", "Bhoomi", "Muskan"],
    "Marks": [85, 90, 78]
})
# DataFrame ->jSON
df.to_json("students.json", orient="records")
# JSON -> DataFrame
df2 = pd.read_json("students.json", orient="records")
print(df2)
# Check equality
print(df.equals(df2))
# %%
#ques 5  write a short html table by hand and load it with read_html.
df = pd.DataFrame({
    "Name": ["Disha", "Bhoomi", "Muskan"],
    "Marks": [85, 90, 78]
})
df.to_html("students.html", index=False)
tables = pd.read_html("students.html")
print(tables[0])
```
